```python
# Count unique students at each stage
created   = events[events['event_name'] == 'practice_test_created']['student_id'].nunique()
submitted = events[events['event_name'] == 'practice_test_submitted']['student_id'].nunique()
cap_start = events[events['event_name'] == 'capsule_started']['student_id'].nunique()
cap_done  = events[events['event_name'] == 'capsule_completed']['student_id'].nunique()

print("=== EXAM HABIT ACTIVATION FUNNEL ===")
print(f"1. Installed:          {total_installs:,}  (100%)")
print(f"2. Test Created:       {created:,}  ({created/total_installs*100:.1f}%)")
print(f"3. Test Submitted:     {submitted:,}  ({submitted/total_installs*100:.1f}%)")
print(f"4. Capsule Started:    {cap_start:,}  ({cap_start/total_installs*100:.1f}%)")
print(f"5. Capsule Completed:  {cap_done:,}  ({cap_done/total_installs*100:.1f}%)")
```

```python
print("=== STEP-BY-STEP DROP-OFF ===")
print(f"Install → Test Created:        {created/total_installs*100:.1f}%  ({total_installs-created:,} lost)")
print(f"Test Created → Submitted:      {submitted/created*100:.1f}%  ({created-submitted:,} lost)")
print(f"Capsule Started → Completed:   {cap_done/cap_start*100:.1f}%  ({cap_start-cap_done:,} lost)")
```

```python
# Step 1: Keep only "learning" events (real learning, not just browsing)
learning_events = events[events['event_name'].isin(['practice_test_submitted', 'capsule_completed'])]

# Step 2: Assign each event to a week
learning_events = learning_events.copy()
learning_events['week'] = learning_events['event_time'].dt.to_period('W').apply(lambda r: r.start_time)

# Step 3: Count unique students per week
wals_by_week = learning_events.groupby('week')['student_id'].nunique()

print("=== WALS (Weekly Active Learning Students) ===")
print(wals_by_week)
```

```python
wals_df = wals_by_week.reset_index()
wals_df.columns = ['week', 'wals']

# pct_change() computes: (this week - last week) / last week * 100
wals_df['wow_growth_%'] = wals_df['wals'].pct_change() * 100

print(wals_df.to_string(index=False))
```

```python
# Step 1: Map each student's install date onto the events table
events['install_date'] = events['student_id'].map(
    students.set_index('student_id')['install_date']
)

# Step 2: Calculate how many days after install each event happened
events['days_since_install'] = (
    events['event_time'].dt.normalize() - events['install_date']
).dt.days

# Step 3: Count students active on exactly Day 1, 7, 30
d1  = events[events['days_since_install'] == 1]['student_id'].nunique()
d7  = events[events['days_since_install'] == 7]['student_id'].nunique()
d30 = events[events['days_since_install'] == 30]['student_id'].nunique()

print("=== RETENTION (% of installs still active) ===")
print(f"D1  (Day 1):  {d1}  →  {d1/total_installs*100:.1f}%")
print(f"D7  (Day 7):  {d7}  →  {d7/total_installs*100:.1f}%")
print(f"D30 (Day 30): {d30}  →  {d30/total_installs*100:.1f}%")
```

```python
# Map city_tier onto events
events['city_tier'] = events['student_id'].map(
    students.set_index('student_id')['city_tier']
)

d7_events = events[events['days_since_install'] == 7]

print("=== D7 RETENTION BY CITY TIER ===")
for tier in ['Tier 1', 'Tier 2', 'Tier 3']:
    # students in this tier
    tier_installs = len(students[students['city_tier'] == tier])
    # students in this tier who were active on D7
    tier_d7 = d7_events[d7_events['city_tier'] == tier]['student_id'].nunique()
    print(f"{tier}: {tier_d7}/{tier_installs} = {tier_d7/tier_installs*100:.1f}%")
```

```python
# Map source onto events
events['source'] = events['student_id'].map(
    students.set_index('student_id')['source']
)

# CPI = Cost per Install (standard metric everyone uses)
campaign_summary = campaigns.groupby('channel').agg(
    total_spend   = ('spend_inr', 'sum'),
    total_installs = ('installs_attributed', 'sum')
).reset_index()

campaign_summary['CPI'] = campaign_summary['total_spend'] / campaign_summary['total_installs'].clip(lower=1)

print("=== STANDARD CPI ===")
print(campaign_summary[['channel','total_spend','total_installs','CPI']].to_string(index=False))
```

```python
#  CPAL = Cost per Active Learner
print("\n=== CPAL - Cost per Active Learner  ===")
learning_events_with_source = events[events['event_name'].isin(['practice_test_submitted','capsule_completed'])]

active_by_source = (
    learning_events_with_source
    .groupby('source')['student_id']
    .nunique()
    .reset_index()
    .rename(columns={'source':'channel', 'student_id':'active_learners'})
)

campaign_summary = campaign_summary.merge(active_by_source, on='channel', how='left')
campaign_summary['CPAL'] = campaign_summary['total_spend'] / campaign_summary['active_learners'].clip(lower=1)

print(campaign_summary[['channel','total_spend','active_learners','CPAL']].to_string(index=False))
```

```python
events['hour'] = events['event_time'].dt.hour

hourly_learning = (
    events[events['event_name'].isin(['practice_test_submitted','capsule_completed'])]
    .groupby('hour')
    .size()
    .reset_index(name='learning_actions')
    .sort_values('hour')
)

print("=== LEARNING ACTIONS BY HOUR ===")
print(hourly_learning.to_string(index=False))

hourly_learning['hour_24'] = hourly_learning['hour'] + 1

fig = px.bar(
    hourly_learning,
    x='hour_24',
    y='learning_actions',
    title='Learning Actions by Hour of Day',
    labels={'hour_24': 'Hour', 'learning_actions': 'Learning Events'},
    color='learning_actions',
    color_continuous_scale='Blues'
)

fig.add_vrect(
    x0=19, x1=24,
    fillcolor='green',
    opacity=0.1,
    annotation_text='Peak Zone (7–11 PM)',
    annotation_position='top left'
)

fig.show()
```
